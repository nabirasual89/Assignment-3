4. What is the purpose of the 'return' statement in a Python function?
 - The return statement in a Python function serves two primary purposes:

Exiting the function:

7. What are the advantages of using generators over regular functions?
 - Generators offer several advantages over regular functions, primarily stemming from their ability to produce values on demand and manage state efficiently. This makes them memory-efficient, ideal for handling large or infinite sequences, and suitable for tasks requiring lazy evaluation
 Here's a more detailed look at the advantages:

Memory Efficiency:

Unlike regular functions that return the entire result at once, generators yield values one by one as needed, reducing memory consumption, especially when dealing with large datasets.

Lazy Evaluation:

Generators compute values only when they are requested, which can improve performance, particularly in scenarios where not all values are required.

Concise Code:

Generators often provide a cleaner and more readable way to create iterators compared to traditional loops.

State Management:

Generators maintain their state between calls, allowing you to resume execution from the last yielded point.

Infinite Sequences:

Generators can be used to create infinite sequences of values, unlike regular functions that need to return a complete result.

Cooperative Multitasking:

Generators can be paused and resumed, facilitating cooperative multitasking where different parts of the code can share the processing load.

In essence, generators are valuable tools for situations where memory efficiency, lazy evaluation, or the ability to manage state between function calls are crucial.

```python
def simple_generator():
    yield 1
    yield 2
    yield 3

# Using the generator
gen = simple_generator()
print(next(gen))  # Output: 1
print(next(gen))  # Output: 2
print(next(gen))  # Output: 3

def fibonacci_generator():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

# Using the generator for an infinite sequence
fib_gen = fibonacci_generator()
for _ in range(10):
    print(next(fib_gen))  # Output: First 10 Fibonacci numbers
```

9. Explain the purpose and usage of the 'map()' funcation in Python.
 - The map() function in Python applies a given function to each item of an iterable (like a list or tuple) and returns a map object (an iterator) containing the results. Its primary purpose is to transform data by applying the same operation to each element in the iterable, avoiding the need for explicit loops.


```python
def square(n):
    return n * n

numbers = [1, 2, 3, 4, 5]
squared_numbers = map(square, numbers)

print(list(squared_numbers))
# Output: [1, 4, 9, 16, 25]
```

In this example, the map() function applies the square() function to each number in the numbers list. The result is a map object, which is then converted to a list to display the squared values. map() can also be used with lambda functions for more concise code.

```python
numbers = [1, 2, 3, 4, 5]
squared_numbers = map(lambda x: x**2, numbers)

print(list(squared_numbers))
# Output: [1, 4, 9, 16, 25]
```

When working with multiple iterables, the function passed to map() must accept the corresponding number of arguments.

```python
list1 = [1, 2, 3]
list2 = [4, 5, 6]
result = map(lambda x, y: x + y, list1, list2)

print(list(result))
# Output: [5, 7, 9]
```

- The map() and filter() functions in Python are both higher-order functions that operate on iterables, but they serve different purposes:
 map():

Applies a given function to each item in an iterable (e.g., a list, tuple) and returns a new iterable (a map object, which can be converted to a list or other sequence type) containing the results. Essentially, it transforms each element of the input iterable.

filter():

Filters elements from an iterable based on a given condition (a function that returns True or False). It returns a new iterable (a filter object) containing only the elements for which the condition is True.


```python
# Example using map()
numbers = [1, 2, 3, 4, 5]
squared_numbers = map(lambda x: x**2, numbers)
print(list(squared_numbers))  # Output: [1, 4, 9, 16, 25]

# Example using filter()
numbers = [1, 2, 3, 4, 5, 6]
even_numbers = filter(lambda x: x % 2 == 0, numbers)
print(list(even_numbers))  # Output: [2, 4, 6]
```

1. Write a Python function that takes a list of numbers as input and returns the sumeven numbers in the list.

```python
def sum_even_numbers(numbers):
    """
    This function takes a list of numbers as input and returns the sum of the even numbers in the list.

    :param numbers: List of integers
    :return: Sum of even integers in the list
    """
    return sum(num for num in numbers if num % 2 == 0)

# Example usage:
numbers = [1, 2, 3, 4, 5, 6]
print(sum_even_numbers(numbers))  # Output: 12
```

This function uses a generator expression to filter out the even numbers and then sums them up using the sum() function. It's concise and efficient!

2. Create a Python function that accepts a string and returns the reverse of that string.

```python


def reverse_string(input_string):
    """
    This function takes a string as input and returns the reversed string.

    Parameters:
    input_string (str): The string to be reversed.

    Returns:
    str: The reversed string.
    """
    return input_string[::-1]

# Example usage:
original_string = "Hello, World!"
reversed_string = reverse_string(original_string)
print(reversed_string)  # Output: !dlroW ,olleH
```

This function uses Python's slicing feature to reverse the string. The [::-1] slice notation means "take the string and step backwards by 1," effectively reversing it.

```python
def square_numbers(numbers):
    """
    This function takes a list of integers and returns a new list containing the squares of each number.

    :param numbers: List of integers
    :return: List of integers where each integer is the square of the corresponding integer in the input list
    """
    return [number ** 2 for number in numbers]

# Example usage:
input_list = [1, 2, 3, 4, 5]
squared_list = square_numbers(input_list)
print(squared_list)  # Output: [1, 4, 9, 16, 25]
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

```python
def is_prime(number):
    """Check if a number is prime within the range of 1 to 200."""
    if number <= 1 or number > 200:
        return False
    if number == 2:
        return True
    if number % 2 == 0:
        return False
    for i in range(3, int(number**0.5) + 1, 2):
        if number % i == 0:
            return False
    return True

# Example usage:
for num in range(1, 201):
    print(f"{num} is {'a prime' if is_prime(num) else 'not a prime'} number.")
```

This function first checks if the number is within the valid range (1 to 200).

It then uses basic rules to determine if the number is prime:


Numbers less than or equal to 1 and greater than 200 are not prime.

The number 2 is prime.

Even numbers greater than 2 are not prime.

For other numbers, it checks divisibility from 3 up to the square root of the number, skipping even numbers.

5. Create an interator class in Python that generates the Fibonacci sequence up to a specified number of terms.

```python


class FibonacciIterator:
    def __init__(self, num_terms):
        self.num_terms = num_terms
        self.current_term = 0
        self.a, self.b = 0, 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current_term >= self.num_terms:
            raise StopIteration
        if self.current_term == 0:
            self.current_term += 1
            return self.a
        elif self.current_term == 1:
            self.current_term += 1
            return self.b
        else:
            self.current_term += 1
            self.a, self.b = self.b, self.a + self.b
            return self.b

# Example usage:
num_terms = 10
fib_iterator = FibonacciIterator(num_terms)
for fib_number in fib_iterator:
    print(fib_number)


```

This class, FibonacciIterator, initializes with the number of terms you want to generate. It implements the __iter__ and __next__ methods to make it an iterator. The __next__ method generates the next Fibonacci number in the sequence until the specified number of terms is reached.



6. Write a genrator function in Python that yields the powers of 2 up to a given exponent.

```python

def powers_of_two(max_exponent):
    """Generator function that yields powers of 2 up to the given exponent."""
    for exponent in range(max_exponent + 1):
        yield 2 ** exponent

# Example usage:
for power in powers_of_two(5):
    print(power)

```

In this example, if you call powers_of_two(5), it will yield the values (2^0, 2^1, 2^2, 2^3, 2^4, \text{and } 2^5).

```python
def read_file_line_by_line(file_path):
    """
    Generator function that reads a file line by line and yields each line as a string.

    :param file_path: Path to the file to be read.
    """
    try:
        with open(file_path, 'r') as file:
            for line in file:
                yield line.strip()
    except FileNotFoundError:
        print(f"The file at {file_path} was not found.")
    except IOError:
        print(f"An error occurred while reading the file at {file_path}.")

# Example usage:
# for line in read_file_line_by_line('example.txt'):
#     print(line)
```

```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

# List of temperatures in Celsius
celsius_temperatures = [0, 20, 37, 100]

# Use map to apply the conversion function to each temperature in the list
fahrenheit_temperatures = list(map(celsius_to_fahrenheit, celsius_temperatures))

# Print the list of temperatures in Fahrenheit
print(fahrenheit_temperatures)
```
